```python
layout = fifi.svg_to_axes.FigureLayout('../../../../Figure/review/odorTraces.svg',
                                           autogenlayers=True, make_mplfigures=True, hide_layers=[])
```

```python
df = pd.read_hdf("~/DataAnalysis/data/TrajectoryInt.h5")
```

```python
source = np.array([[0,0]])
odor_position = np.array([[df.xsrc[i],df.ysrc[i]] for i in range (len(df))]) 

distance_from_source = np.array([cdist(odor_position,source)]).flatten()
df['distance_from_source'] = distance_from_source
```

```python
ax=layout.axes[('fig1','ax1')]
# mean_avg_distance = 18.73m
lw=1
msize=3

ax.plot(df.time,df.odor, c='b', linewidth=lw,rasterized=True)

ax.grid(False)
ax.set_xlim(140,150)
ax.set_rasterization_zorder(-20)

mpl_functions.adjust_spines(ax,['left'],
                            spine_locations={'left':4,'bottom':5}, 
                            smart_bounds=True,
                            xticks=[140,150],
                            yticks = [0,10],
                            linewidth=0.5)
ax.set_xticklabels([])
# ax.set_yticklabels([])

ax.set_ylabel('Odor, a.u.')
ax.yaxis.set_label_coords(-0.04, 0.5)


# ax.set_xlabel('10 secs')
# ax.xaxis.set_label_coords(0.5, -0.2)

figurefirst.mpl_functions.set_fontsize(ax, FS)
layout.append_figure_to_layer(layout.figures['fig1'], 'fig1', cleartarget=True)
layout.write_svg('../../../../Figure/review/odorTraces.svg')
```
